```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("brfss2022_data_wrangling_output.csv")
print("Shape:", df.shape)
df.head()

```

```python
df = df.astype("category")
```

```python
#Target Variable Distribution
df["heart_disease"].value_counts().plot(kind="bar", color=["green", "red"])
plt.title("Heart Disease Distribution")
plt.xlabel("Heart Disease")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()



```

```python
#Heart Disease vs Age Category
crosstab = pd.crosstab(df["age_category"], df["heart_disease"])
crosstab.plot(kind="bar", stacked=True, color=["green", "red"], figsize=(12,6))
plt.title("Heart Disease vs Age Category")
plt.xlabel("Age Category")
plt.ylabel("Count")
plt.legend(title="Heart Disease")
plt.xticks(rotation=45)
plt.show()


```

```python
#Percentage Distribution (Optional)
crosstab_perc = crosstab.div(crosstab.sum(axis=1), axis=0) * 100
crosstab_perc.plot(kind="bar", stacked=True, color=["green", "red"], figsize=(12,6))
plt.title("Heart Disease vs Age Category (Percentage)")
plt.xlabel("Age Category")
plt.ylabel("Percentage (%)")
plt.legend(title="Heart Disease")
plt.xticks(rotation=45)
plt.show()

```
